```python
import pickle
import os
from collections import Counter
import matplotlib.pyplot as plt
```

```python
path = "data/"
configs = ["MO-Swimmer-v2", "MO-Walker2d-v2", "MO-Ant-v2", "MO-HalfCheetah-v2", "MO-Hopper-v2"]

for config in configs:
    for quality in ["expert", "amateur"]:
        for dist in ["uniform", "wide", "narrow"]:
            content_path = f"{path}{config}/{config}_50000_{quality}_{dist}.pkl"

            if os.path.isfile(content_path):
                content = pickle.load(open(content_path, "rb"))
                preferences = []
                item_count = 0

                for item in content:
                    preference_weights = item["preference"][0]
                    preferences.extend(preference_weights)
                    item_count += 1

                print(f"\n{config} {quality} {dist}:")
                print(f"Amount of items: {item_count}")
                print(f"Number of preference weights: {len(preferences)}")

                fig = plt.figure(figsize=(10, 7))
                plt.hist(preferences, bins=40)
                plt.title(f"{config} {quality} {dist}")
                fig.savefig(f"data_figs/{config}_{quality}_{dist}")
                plt.close(fig)
```
